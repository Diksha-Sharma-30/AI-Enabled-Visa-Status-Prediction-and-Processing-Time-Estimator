```python
columns = [
'CASE_STATUS',
'RECEIVED_DATE',
'DECISION_DATE',
'VISA_CLASS',
'JOB_TITLE',
'FULL_TIME_POSITION',
'EMPLOYER_STATE',
'WORKSITE_STATE',
'WAGE_RATE_OF_PAY_FROM',
'PREVAILING_WAGE'
]

df = df[columns]
df.head()
```

```python
df = df.dropna()

print("After removing missing values:")
print(df.shape)
```

```python
df['RECEIVED_DATE'] = pd.to_datetime(df['RECEIVED_DATE'], errors='coerce')
df['DECISION_DATE'] = pd.to_datetime(df['DECISION_DATE'], errors='coerce')
```

```python
print(df[['RECEIVED_DATE', 'DECISION_DATE']].dtypes)
```

```python
df['processing_days'] = (df['DECISION_DATE'] - df['RECEIVED_DATE']).dt.days
df = df[df['processing_days'] >= 0]
df.head()
```

```python
df[['RECEIVED_DATE', 'DECISION_DATE', 'processing_days']]
```

```python
encoder = LabelEncoder()

categorical_cols = [
'CASE_STATUS',
'VISA_CLASS',
'JOB_TITLE',
'FULL_TIME_POSITION',
'EMPLOYER_STATE',
'WORKSITE_STATE'
]

for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col])
```

```python
scaler = StandardScaler()

df[['WAGE_RATE_OF_PAY_FROM','PREVAILING_WAGE']] = scaler.fit_transform(
    df[['WAGE_RATE_OF_PAY_FROM','PREVAILING_WAGE']]
)
```

```python
df = df.drop(columns=['RECEIVED_DATE','DECISION_DATE'])
df.head()
```

```python
df.to_csv("cleaned_visa_dataset.csv", index=False)
```
