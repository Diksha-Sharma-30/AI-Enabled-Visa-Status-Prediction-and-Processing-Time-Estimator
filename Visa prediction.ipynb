```python
import pandas as pd
```

```python
import numpy as np
```

```python
df=pd.read_csv("dataset.csv")
df.shape
df.head()
```

```python
print("Shape:", df.shape)
df.head()
```

```python
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(include='object').columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[cat_cols] = df[cat_cols].apply(lambda x: x.fillna(x.mode()[0]))

print("\nMissing Values After:\n")
print(df.isnull().sum())
```

```python
print(df.columns)
```

```python
columns = [
'CASE_STATUS',
'RECEIVED_DATE',
'DECISION_DATE',
'VISA_CLASS',
'JOB_TITLE',
'FULL_TIME_POSITION',
'EMPLOYER_STATE',
'WORKSITE_STATE',
'WAGE_RATE_OF_PAY_FROM',
'PREVAILING_WAGE'
]

df = df[columns]
df.head()
```

```python
df['RECEIVED_DATE'] = pd.to_datetime(df['RECEIVED_DATE'], errors='coerce')
df['DECISION_DATE'] = pd.to_datetime(df['DECISION_DATE'], errors='coerce')
```

```python
print(df[['RECEIVED_DATE', 'DECISION_DATE']].dtypes)
```

```python
df['processing_days'] = (df['DECISION_DATE'] - df['RECEIVED_DATE']).dt.days
```

```python
df[['RECEIVED_DATE', 'DECISION_DATE', 'processing_days']]
```

```python
df.to_csv("cleaned_visa_dataset.csv", index=False)
```
